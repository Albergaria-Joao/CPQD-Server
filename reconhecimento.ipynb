```python
import cv2
import pytesseract

# Se estiver no Windows e o tesseract não estiver no PATH, descomente:
# pytesseract.pytesseract.tesseract_cmd = r"C:\Program Files\Tesseract-OCR\tesseract.exe"

img = cv2.imread("Placa\test\images\placa_000451.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# binarização simples pra tentar melhorar OCR
_, thresh = cv2.threshold(gray, 150, 255, cv2.THRESH_BINARY)

texto = pytesseract.image_to_string(thresh, lang="por")
print(texto)
```
